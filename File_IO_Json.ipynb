````python
import json
def save_llm_response_log(query,response,log_file = 'llm_response.log'):
    with open(log_file,'a',encoding = 'utf-8') as f:
        entry = json.dumps({'query' : query,'response':response})
        f.write(entry + '\n')


````

````python

save_llm_response_log('Where is my order?', 'Your order ORD-3042 is In Transit.')
````

````python
save_llm_response_log('Can I return this?', 'Returns accepted within 30 days.')
````

````python
import json
with open('llm_response.log', 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        record = json.loads(line.strip())
        print(f'  [{i+1}] Q: {record["query"]}')
        print(f'       A: {record["response"]}')
````

````python
import json
def parse_llm_response(raw_response):
    cleaned = raw_response.strip()

    if cleaned.startswith('```'):
        cleaned = cleaned.split('\n',1)[-1] # split on first line and take the second element first is ```
        cleaned = cleaned.split('```',1)[0] # split json and '```' and takes first element

    return json.loads(cleaned)
        
        

````

````python
fenced = '```json\n{"intent": "CANCEL", "urgency": "high"}\n```'

parsed = parse_llm_response(fenced)
print(f"Before_parsed:", fenced)
print(f"After_parsed:", parsed)
````

````python
raw = '{"category": "TRACK_ORDER", "confidence": "high"}'
````

````python
print(f"Before parsed:", raw)
parsed_new = parse_llm_response(raw)
print(f"After parsed:", parsed_new['category'])
````
