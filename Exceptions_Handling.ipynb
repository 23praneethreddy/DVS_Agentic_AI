```python
def safe_to_int(value):
    try:
        return int(value)
    except ValueError:
        print(f" ValueError : '{value}' cannot be int")
        return None

print(safe_to_int('100'))
print(safe_to_int('3.14'))
print(safe_to_int('abc'))
        
    
```

```python
def get_customer_details(customer:dict,key: str):
    try:
        return customer[key].strip().title()
    except KeyError:
        print(f"KeyError: key '{key}' is not a right key")
        return 'Unknown'
    except AttributeError:
        print(f"AttributeError: value for '{key}' is None")
        return None



customer = {'customer_id': '1001', 'name': 'prudhvi akella', 'email': None}

print(get_customer_details(customer,'name'))
print(get_customer_details(customer,'email'))
print(get_customer_details(customer,'city'))
```

```python
import traceback
def validate_customer_id(customer_id:str)->int:

    try:
        cid = int(customer_id)   # converting customer_id from string to integer
    except ValueError as e:
        raise ValueError(f"customer_id must be number.got {customer_id}") from e
    if cid < 0 :
        raise ValueError(f"customer_id must be number.got {cid}")
    return cid
    
# validate_customer_id('abc')
try:
    validate_customer_id('-5')
except ValueError as e:
    traceback.print_exc()
except Exception as e:
    print(f"Caught: {e}")

try:
    validate_customer_id('abc')
except ValueError as e:
    print(f"Caught: {e}")
except Exception as e:
    print(f"Caught: {e}")

print('X')

    
```

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as e:
        raise ZeroDivisionError("b cannot be zero") from e
    return result

divide(10, 0)
```

```python
#raw_id --> convert to int-->customer id and deatils --> cuatomer name and id print or catch the error raise in parsed function in 


def parsed_customer_id(raw_id):
    try:
        cid = int(raw_id)
    except ValueError as e:
        raise ValueError(f"customer_id is not number. Got {raw_id}") from e
    if cid<0:
        raise ValueError(f"customer_id is not number. Got {cid}") 
    return cid

def load_customer(raw_id):
    customer_id = parsed_customer_id(raw_id)
    return {'customer_id' : customer_id,'name' :'Saanvi','salary':120000}

def build_customer_prompt(raw_id):
    try:
       customer = load_customer(raw_id)
       return f"customer_name':{customer['name']},(id={customer['customer_id']})"
    except ValueError as e:
        return f"Could not build prompt — bad customer ID: {e}"

print(build_customer_prompt('-5'))
        
```
